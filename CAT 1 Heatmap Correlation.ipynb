```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
url= "https://raw.githubusercontent.com/emmanuelyardamah/Data-Warehouse-and-Data-Mining/refs/heads/main/cleaned_uganda_crashes_2011_2020.csv"
df= pd.read_csv(url)
```

```python
# Select our 4 variables
cols = ['Population_Behavior','Road_Environment','Motor_Vehicle_Condition','Total_Road_Traffic_Accidents']

# Filter columns that are actually present in the dataframe to avoid KeyErrors
existing_cols = [col for col in cols if col in df.columns]

if len(existing_cols) > 0:
# Explicitly convert selected columns to numeric to avoid NaN correlations due to object dtype
    for col in existing_cols:
        df[col] = pd.to_numeric(df[col], errors='coerce')

    corr = df[existing_cols].corr()

    # Heatmap
    plt.figure(figsize=(9,6))
    sns.heatmap(corr, annot=True, cmap='cividis', vmin=-1, vmax=1, fmt='.3f')
    plt.title(' Our Correlation Heatmap - Causes of Road Crashes Uganda 2011-2020')
    plt.tight_layout()
    plt.savefig('heatmap_uganda_crashes.png')
    plt.show()
else:
    print(f"None of the expected columns {cols} were found in the DataFrame. Available columns: {list(df.columns)}")
```
