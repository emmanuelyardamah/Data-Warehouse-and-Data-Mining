```python
import pandas as pd
url = "https://raw.githubusercontent.com/emmanuelyardamah/Data-Warehouse-and-Data-Mining/refs/heads/main/retail_orders_dirty.csv"
df = pd.read_csv(url)
df.head()
```

```python
df.isnull().sum()
```

```python
# Option 1: Drop all rows containing any null values
df_cleaned_drop = df.dropna()
print(f"Original shape: {df.shape}")
print(f"Shape after dropping nulls: {df_cleaned_drop.shape}")
```

```python
# Option 2: Strip whitespaces from column names first, then fill null values
df_cleaned_fill = df.copy()
df_cleaned_fill.columns = df_cleaned_fill.columns.str.strip()

df_cleaned_fill['Gender'] = df_cleaned_fill['Gender'].fillna('Unknown')
df_cleaned_fill['City'] = df_cleaned_fill['City'].fillna('Unknown')
df_cleaned_fill['AGE'] = df_cleaned_fill['AGE'].fillna(df_cleaned_fill['AGE'].median())

display(df_cleaned_fill.isnull().sum())
```
